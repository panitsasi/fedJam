```python
import pandas as pd
import matplotlib.pyplot as plt
import IPython.display as display
import time

CSV_FILE = 'signal_measurements.csv'

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))
fig.tight_layout(pad=4.0)

while True:
    try:
        df = pd.read_csv(CSV_FILE)
        df = df[(df['Noise'] > -110) & (df['Noise'] < -70)]
        df = df[(df['SNR'] > 40) & (df['SNR'] < 70)]


        df['Time'] = pd.to_datetime(df['Time'])
        df = df[['Time', 'Noise', 'SNR']]
        df['Elapsed'] = (df['Time'] - df['Time'].iloc[0]).dt.total_seconds()

        ax1.clear()
        ax2.clear()

        ax1.plot(df['Elapsed'], df['SNR'], color='green')
        ax1.set_title('SNR (dB) Over Time')
        ax1.set_ylabel('SNR (dB)')
        ax1.grid(True)

        ax2.plot(df['Elapsed'], df['Noise'], color='orange')
        ax2.set_title('Noise Floor (dBm) Over Time')
        ax2.set_xlabel('Time (s)')
        ax2.set_ylabel('Noise (dBm)')
        ax2.grid(True)

        display.clear_output(wait=True)
        display.display(fig)

        time.sleep(1)

    except Exception as e:
        print(f"Error: {e}")
        break
```
