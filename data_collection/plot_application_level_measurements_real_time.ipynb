```python
import pandas as pd
import matplotlib.pyplot as plt
import IPython.display as display
import time

CSV_FILE = 'application_level_measurements.csv'



# Create one figure with 3 subplots
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(12, 14))
fig.tight_layout(pad=5.0)

while True:
    try:
        df = pd.read_csv(CSV_FILE)
        df = df.iloc[1:] 

        df['Time'] = pd.to_datetime(df['Time'])
        df['Elapsed'] = (df['Time'] - df['Time'].iloc[0]).dt.total_seconds()

        # Clear old plots
        ax1.clear()
        ax2.clear()
        ax3.clear()

        # Plot 1: Latency
        ax1.plot(df['Elapsed'], df['Latency'], color='green')
        ax1.set_title('Latency Over Time')
        ax1.set_xlabel('Time (s)')
        ax1.set_ylabel('Latency (ms)')
        ax1.grid(True)

        # Plot 2: Jitter
        ax2.plot(df['Elapsed'], df['Jitter'], color='blue')
        ax2.set_title('Jitter Over Time')
        ax2.set_xlabel('Time (s)')
        ax2.set_ylabel('Jitter (ms)')
        ax2.grid(True)

        # Plot 3: Packet Loss
        ax3.plot(df['Elapsed'], df['Packet Loss Count'], color='red', marker='o')
        ax3.set_title('Packet Loss Over Time')
        ax3.set_xlabel('Time (s)')
        ax3.set_ylabel('Lost Packets')
        ax3.grid(True)


        # Update display
        display.clear_output(wait=True)
        display.display(fig)

        time.sleep(1)

    except Exception as e:
        print(f"Error: {e}")
        break
```
